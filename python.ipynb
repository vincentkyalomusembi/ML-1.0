DATA VISUALIZATION

```python
df.columns
```

```python
score_cols = ['math_score', 'history_score', 'physics_score',
       'chemistry_score', 'biology_score', 'english_score', 'geography_score']


df["average_score"] = df[score_cols].mean(axis=1)#working across  row  (a student at a time)
df["total_score"] = df[score_cols].sum(axis=1)
print(df[["average_score", "total_score"]].head(30))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

**One column at a time (distribution)**

```python
sns.histplot(df["average_score"], bins=10, kde=True)
plt.title("Distribution of Average Scores")
plt.xlabel("Average Score")
plt.ylabel("Frequency")
plt.show()
```

```python
df[score_cols].hist(bins=20, figsize=(10, 8))
plt.xlabel("Score")
plt.ylabel("number of Students")
plt.tight_layout()#spacing between plots
plt.show()
```

Box plot for each subject - shows the middle score, spread & unusual students

```python
sns.boxplot(data=df[score_cols])
plt.title("Boxplot of Scores by Subject")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(data=df, x="gender")
plt.title("student by Gender")
plt.show()
```

```python
df["career_aspiration"].value_counts().plot(kind="bar")
plt.title("Career Aspirations of Students")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))    # 1 row, 2 charts
sns.histplot(df["weekly_self_study_hours"], bins=20, ax=axes[0])
axes[0].set_title("Weekly Study Hours")
sns.histplot(df["absence_days"], bins=15, ax=axes[1])
axes[1].set_title("Absence Days")
plt.show()
```

**Two columns together (relationships)**

```python
sns.scatterplot(data=df, x="weekly_self_study_hours", y="average_score", alpha=0.4)

sns.regplot(data=df, x="weekly_self_study_hours", y="average_score",
            scatter=False, color="red")
plt.title("Study Hours vs Average Score")
plt.show()
```

```python
# WHAT: Same chart, but for absence days.
# WHY: Shows if missing school goes with lower scores (a line going DOWN).
sns.scatterplot(data=df, x="absence_days", y="average_score", alpha=0.4)
sns.regplot(data=df, x="absence_days", y="average_score", scatter=False, color="red")
plt.title("Absence Days vs Average Score")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.boxplot(data=df, x="part_time_job", y="average_score", ax=axes[0])
sns.boxplot(data=df, x="extracurricular_activities", y="average_score", ax=axes[1])
sns.boxplot(data=df, x="gender", y="average_score", ax=axes[2])
plt.show()
```
