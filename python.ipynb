DATA VISUALIZATION

```python
df.columns
```

```python
score_cols = ['math_score', 'history_score', 'physics_score',
       'chemistry_score', 'biology_score', 'english_score', 'geography_score']


df["average_score"] = df[score_cols].mean(axis=1)#working across  row  (a student at a time)
df["total_score"] = df[score_cols].sum(axis=1)
print(df[["average_score", "total_score"]].head(20))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.histplot(df["average_score"], bins=10, kde=True)
plt.title("Distribution of Average Scores")
plt.xlabel("Average Score")
plt.ylabel("Frequency")
plt.show()
```

```python
df[score_cols].hist(bins=20, figsize=(10, 8))
plt.xlabel("Score")
plt.ylabel("number of Students")
plt.tight_layout()#spacing between plots
plt.show()
```
